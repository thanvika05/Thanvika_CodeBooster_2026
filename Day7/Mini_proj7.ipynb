```python
documents = [
    "Arun is a CSE student with a CGPA of 8.5",
    "Priya studies ECE and has a CGPA of 9.1",
    "Karthik is an IT student interested in web development",
    "Divya is a CSE student who likes machine learning",
    "Rahul studies EEE and participates in robotics competitions"
]

print(documents)
```

```python
query = "CSE"

results = []

for doc in documents:
    if query.lower() in doc.lower():
        results.append(doc)

print("Keyword Search Results:")
for r in results:
    print(r)
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

doc_vectors = model.encode(documents)

print(doc_vectors.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

query = "computer science student"

query_vector = model.encode([query])

scores = cosine_similarity(query_vector, doc_vectors)[0]

print(scores)
```

```python
top_indices = scores.argsort()[::-1]

print("Semantic Search Results:\n")

for idx in top_indices:
    print(f"Score: {scores[idx]:.4f}")
    print(documents[idx])
    print()
```

```python
import numpy as np

query = "student interested in artificial intelligence"

query_vector = model.encode([query])[0]

similarities = np.dot(doc_vectors, query_vector) / (
    np.linalg.norm(doc_vectors, axis=1) *
    np.linalg.norm(query_vector)
)

best_match = np.argmax(similarities)

print("Best Match:")
print(documents[best_match])
```
